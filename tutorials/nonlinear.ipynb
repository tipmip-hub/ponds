# Generating a set of nonlinear timeseries that include shifts

With ponds, this is done by *first* creating a regular dataset of timeseries that will serve as the forcing and *then* feed that dataset as a whole into a system that would show nonlinear dynamics. 

We can tweak PONDS core methods a bit to serve this purpose, and finally use `PONDS.apply_nonlinear_transformation()` to generate the nonlinear dataset.

## 1. Create a ponds object and choose a nonlinear system

Note: Yet, only the Lorenz system is implemented, and can be forced through the beta parameter.

```python
pond = PONDS(
    lat_size=20,
    lon_size=40,
    time_size=2000
)


# instantiate a nonlinear system and choose forced parameter
system = Lorenz(forced_parameter='beta')
```

## 2. Set the ground state 

In our ponds object, set ground state where all added shifts would depart from. This can be realized by retrieving the respective parameter value from our system, and then adding that as a constant to our dataset using `Trend`:

```python
ground_state = 'p1'                     # periodic state
ground_value = system.get_parameter_value(state=ground_state)

pond.add_perturbation(                  # better call this "add_background"?
    perturbation_method=Trend(
        slope=0,
        intercept=ground_value,
    )
)
```

## 3. Add some shifts to the dataset

For each shift, choose what state the system should shift to from the states stored in `system.get_parameter_value()`.

```python
# add shift-blob with custom parameters
shifted_state = 'c1'                    # chaotic state
shifted_value = system.get_parameter_value(state=shifted_state)

shift_dict = {
    "shift_steepness": 0.2,
    "shift_magnitude": shifted_value - ground_value,
}
pond.add_shift(
    shape_method=BLOB(random_seed=4),
    shift_method=SIGMOID(**shift_dict),
)


# add another shift-blob with custom parameters
shifted_state = 'c2'                    # different chaotic state
shifted_value = system.get_parameter_value(state=shifted_state)

shift_dict = {
    "shift_steepness": 0.2,
    "shift_magnitude": shifted_value - ground_value,
}
pond.add_shift(
    shape_method=BLOB(random_seed=2),
    shift_method=SIGMOID(**shift_dict),
)
```

```python
# look at the generated synthetic data
pond.worldmap(show_cells=True)
```

## 4. Feed dataset into the chosen nonlinear System

Now our dataset of shifting timeseries can be transformed by a nonlinear system such as the Lorenz system to obtain a set of nonlinear timeseries. This is done with `PONDS.apply_nonlinear_transformation()`.

```python
pond.apply_nonlinear_transformation(
    system=system,
    dt=5e-4,
    sampling_factor=500,
)
```

see how a new variable was added to the ponds object: `nonlinear_ts`

```python
pond.data
```

## Plot

select a shifting timeseries and plot

```python
lat = 18
lon = 22

plt.figure(figsize=(16, 2))
pond.data.nonlinear_ts.isel(lat=lat, lon=lon).plot(lw=.4)

plt.figure(figsize=(16, 2))
pond.data.ts.isel(lat=lat, lon=lon).plot(lw=.5)
```

plot an array of shifting and non-shifting timeseries

```python
plt.figure(figsize=(16, 2))
pond.data.nonlinear_ts.isel(lat=lat, lon=range(40)).plot.line(x='time', lw=.4)
plt.legend().set_visible(False)

plt.figure(figsize=(16, 2))
pond.data.ts.isel(lat=lat, lon=range(40)).plot.line(x='time', lw=.5)
plt.legend().set_visible(False)
```
